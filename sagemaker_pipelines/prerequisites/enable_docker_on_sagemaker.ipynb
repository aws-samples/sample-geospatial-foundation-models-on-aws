# Set up Docker on SageMaker AI

## 1. Enable Docker on your SageMaker AI Domain (one time)

````python
import boto3

sagemaker = boto3.client('sagemaker')
# List all domains
response = sagemaker.list_domains()
domain_id = response['Domains'][0]['DomainId']

print(f"Run the following command in CloudShell in your selected region (e.g., us-eas-1) to ensure Docker is available in SageMaker:\n\naws sagemaker update-domain --domain-id {domain_id} --domain-settings-for-update '{{\"DockerSettings\": {{\"EnableDockerAccess\": \"ENABLED\"}}}}'")
````

> After you have run the the command above, save all your work and shut down, then restart the Jupyter Lab Space. This will propagate the setting change.
